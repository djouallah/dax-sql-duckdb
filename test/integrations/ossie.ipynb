```python
import duckdb

con = duckdb.connect()
con.sql("INSTALL dax_sql FROM community;LOAD dax_sql;")
```

```python
# A Fabric Python notebook signs in to OneLake by itself. On a laptop (az login) or with OIDC,
# the extension takes its token from a credential_chain secret.
try:
    import notebookutils  # only in Fabric
except ImportError:
    con.execute("CREATE OR REPLACE SECRET onelake (TYPE azure, PROVIDER credential_chain)").fetchall()
```

```python


con.execute(f"""
CREATE OR REPLACE SEMANTIC VIEW nem
    FROM 'abfss://power@onelake.dfs.fabric.microsoft.com/nem.Lakehouse/Context/Semantic Views/nem/ossie.yaml';

""").fetchall()
```

```python
con.sql(f"""
SHOW SEMANTIC VIEWS
""").show(max_width=120)
```

```python
con.execute(f"""
describe nem
""").fetchdf()
```

```python
con.execute(f"""
SELECT dim_region.Region, AGG("Registered units") AS units, AGG("Registered MW") AS mw
FROM nem
GROUP BY ALL
ORDER BY ALL;
""").fetchdf()
```
